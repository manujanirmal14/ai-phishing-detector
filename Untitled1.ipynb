```python
import ipywidgets as widgets
from IPython.display import display, clear_output

SUSPICIOUS_WORDS = ["login", "signin", "secure", "verify", "update",
                    "account", "bank", "confirm", "paypal", "password"]

def smart_check(url):
    feats = extract_features(url).iloc[0]
    if not url.startswith("http"):
        url = "http://" + url
    host = urlparse(url).netloc.lower()
    reasons = []

    # සරල නීති (Rules)
    if feats['having_IPhaving_IP_Address'] == -1:
        reasons.append("Uses an IP address instead of a domain name")
    if feats['having_At_Symbol'] == -1:
        reasons.append("Contains '@' symbol")

    # අලුත් rule: domain නමේ සැක සහිත වචන
    found = [w for w in SUSPICIOUS_WORDS if w in host]
    if len(found) >= 2 or (len(found) >= 1 and "-" in host):
        reasons.append("Domain contains suspicious words: " + ", ".join(found))

    # AI එකේ තීරණය
    if model2.predict(extract_features(url))[0] == -1:
        reasons.append("AI model flagged this URL as suspicious")

    if reasons:
        return "⚠️ PHISHING", reasons
    return "✅ SAFE", reasons


# ---- Check කරන පෙට්ටිය (Product Demo) ----
box = widgets.Text(placeholder="Paste a link here...", description="URL:",
                   layout=widgets.Layout(width="600px"))
button = widgets.Button(description="Check", button_style="primary")
out = widgets.Output()

def on_click(b):
    with out:
        clear_output()
        label, why = smart_check(box.value.strip())
        print("Result:", label)
        for r in why:
            print(" - Reason:", r)

button.on_click(on_click)
print("🛡️ AI Phishing Detector")
display(box, button, out)
```
